# 1IEP alternative-conformer challenge — 2026-10-08

Prespecified +/-60 degree central-torsion controls; saved observations only. Evaluate against the original crystallographic geometry. See [workflow](1iep_conformer_workflow.md) for producer boundaries and limits.

```python
import gzip, json
from devtools.qualify_1iep_conformers import OUTPUT, load_baseline, comparison_rows
from devtools.qualify_1iep_root_order import load_baseline as load_representation
from devtools.qualify_1iep_representation import sha
record = json.loads(gzip.decompress(OUTPUT.read_bytes()))
baseline, representation = load_baseline(), load_representation()
assert record['comparison_rows'] == comparison_rows(record, baseline, representation)
assert len(record['new_runs']) == 24
for case in record['cases'].values():
    for control in case['controls'].values():
        assert sha(control['pdbqt'].encode()) == control['sha256']
print('24 new observations; 12 authenticated historical cells with original producers.')
print('Native producer:', record['baseline']['native_producer'])
print('Fixed-first producer:', record['baseline']['fixed_first_producer'])
```

```python
from IPython.display import display, Markdown
rows = record['comparison_rows']
table = ['| Geometry | Order | Effort | Seed | Poses | First heavy RMSD (A) | Closest (A) | First score (kcal/mol) |', '| --- | --- | ---: | ---: | ---: | ---: | ---: | ---: |']
for r in rows:
    table.append(f"| {r['geometry']} | {r['order']} | {r['exhaustiveness']} | {r['seed']} | {r['poses']} | {r['first_heavy_rmsd_angstrom']:.4f} | {r['closest_heavy_rmsd_angstrom']:.4f} | {r['first_vina_score_kcal_mol']:.3f} |")
display(Markdown('\n'.join(table)))
for geometry in ('bound_like', 'plus60', 'minus60'):
    for order in ('native', 'first_fixed'):
        arm = [r for r in rows if r['geometry']==geometry and r['order']==order]
        print(geometry, order, 'first/set near-native:', sum(r['first_pose_recovered'] for r in arm), sum(r['returned_set_recovered'] for r in arm), '/', len(arm))
```

```python
import numpy as np
import pyunitwizard as puw
import dockingmt as dmt
run = record['new_runs'][0]
original_xyz = np.array(record['audit']['source_snapshot']['structures']['coordinates'])[0] * 10
indices = run['pdbqt_to_source_atom_indices']
heavy = run['heavy_pdbqt_indices']
with puw.context(standard_units=['pm','fs','K','mole','dalton','e','kJ/mol','radians']):
    result = dmt.DockingResult.from_dict(run['result'])
    report = dmt.evaluate_redocking(result, puw.quantity(original_xyz[indices], 'angstrom'), rmsd_cutoff=puw.quantity(250, 'pm'))
    assert dmt.verify_captured_inputs(result.provenance['backend_artifacts'])
    for pose, row, saved, metric in zip(result.poses, report['poses'], run['evaluation']['poses'], run['heavy_atom_metrics'], strict=True):
        squared = np.sum((puw.get_value(pose.coordinates, to_unit='angstrom')-original_xyz[indices])**2, axis=-1)
        assert report['criterion'] == run['evaluation']['criterion']
        rmsd = row['rmsd']
        assert abs(float(rmsd)-saved['rmsd']) < 1e-10
        assert abs(float(rmsd)-np.sqrt(squared.mean())) < 1e-10
        assert abs(metric['heavy_atom_positional_rmsd_angstrom']-np.sqrt(squared[heavy].mean())) < 1e-10
print('Saved result re-evaluated against ORIGINAL geometry under pm/fs; no searches repeated.')
print('Unminimized controls, one complex and external unassessed receptor; no general ROOT/conformer policy.')
```
