# Saved E4 carrier: controlled consumer receiving

Unexecuted qualification notebook. One process receives one profile (`normal` or
`pm_fs`) from the registered BINDINGMT_E4_INPUTS manifest. The destination is each exact
saved carrier3238/one structure, including its two complete named analyses.
The original five-pose result remains intact. No scene2024 reconstruction,
one-pose result, new addon API, numerical detector/metric or missing-state codec.

Public registry clear disables unrelated addon discovery only in this fresh dedicated
qualification process; its intentional registry state is retired with the process.
This is not a recommendation for user applications and does not admit the DMT addon;
views are created without display and closed finally. Generic Viewer association
is separate from BindingMT addon navigation and real comm/canvas acceptance.
Queries inspect saved data; typed InteractionsDict/H5MSM provides full persistence.
Diagnostic JSON is never decoded into a molecular system or complete analysis.

Input SHA/physical/unit/state/axis and source nonmutation checks precede alignment
declarations. Expected refusals preserve names and data. Counts are explicit direct
requests; observed aliases include internal reads/getters. Worker supplies
registered_installed_check and isolated consumer/provider/Viewer artifact checks.
CPU0/nice10/one worker; per profile CPU45s/wall60s, RSS1GiB/AS2GiB/reserve8GiB.
Both profiles and detached receiving remain necessary; this baseline does not
accept the other four poses, arbitrary projection, whole#80 or canvas.


```python
import gzip
import hashlib
import json
import math
import os
import sys
from datetime import datetime, timezone
from functools import wraps
from pathlib import Path

root = Path(os.environ['BINDINGMT_E4_ROOT']).resolve()
output = Path(os.environ['BINDINGMT_E4_OUTPUT']).resolve()
profile = os.environ['BINDINGMT_E4_PROFILE']
assert profile in ('normal', 'pm_fs') and not output.exists()
assert not root.is_relative_to(output)
output.mkdir(parents=True, exist_ok=False)
budgets = dict(
    registry_clear=1,
    original_result_read=1,
    carrier_reads=2,
    copies=2,
    viewer_factory=2,
    viewer_load=2,
    independent_layer_reads=2,
    typed_exports=64,
    analysis_queries=4,
    analysis_pages=4,
    query_inspections=4,
    analysis_get=16,
    attach=12,
    viewer_layer_load=4,
    wrong_axes_remap=2,
    invalid_axes_remap=2,
    viewer_save=2,
    persistence_layer_reads=2,
    viewer_close=2,
    native_snapshots=8,
    source_selections=2,
    consumer_preflight_refusal=1,
)
counts = dict.fromkeys(budgets, 0)
forbidden_attempts = []
observed = dict(public_convert=0, public_layer_reads=0, public_get=0)
ledger = [
    dict(case=name, status='not_attempted')
    for name in ('retained_pose', 'translated_geometry_control')
]
current = None
stage = 'custody'
patches = []
receipts = []
views = []
view_cleanup = []


def sha(path):
    with path.open('rb') as stream:
        return hashlib.file_digest(stream, 'sha256').hexdigest()


def save_json(path, value):
    with path.open('x') as stream:
        json.dump(value, stream, indent=2, allow_nan=False)
        stream.write('\n')


def journal(operation, **fields):
    with (output / 'journal.jsonl').open('a') as stream:
        stream.write(
            json.dumps(
                dict(
                    operation=operation,
                    stage=stage,
                    case=current,
                    counts=counts.copy(),
                    at_utc=datetime.now(timezone.utc).isoformat(),
                    **fields,
                ),
                allow_nan=False,
            )
            + '\n'
        )


def save_ledger():
    (output / 'ledger.json').write_text(
        json.dumps(ledger, indent=2, allow_nan=False) + '\n'
    )


save_ledger()
try:
    inputs_path = Path(os.environ['BINDINGMT_E4_INPUTS']).resolve()
    inputs_bytes = inputs_path.read_bytes()
    save_json(output / 'inputs_manifest.json', json.loads(inputs_bytes))
    inputs_manifest = json.loads(inputs_bytes)
    inputs_manifest_before = json.loads(inputs_bytes)
    assert inputs_manifest['profile'] == profile
    assert inputs_manifest['schema'] == 'bindingmt.e4_saved_carrier_inputs@1'
    assert inputs_manifest['owner'] == 'uibcdf/bindingmt#80'
    assert (
        inputs_manifest['original_source'] == '5952053328aac6f737e22906e984c0f9eb938dd7'
    )
    entries = inputs_manifest['inputs']
    assert set(entries) == {
        'result',
        'association',
        'retained_pose',
        'translated_geometry_control',
    }
    base = f'devguide/validation/data/3ptb_interactions_hosted_5952053/{profile}/interactions/'
    paths = dict(
        result='devguide/validation/data/human_3ptb/native_result_2026-10-20.json.gz',
        association=base + 'association.json.gz',
        retained_pose=base + 'retained_pose.h5msm.gz',
        translated_geometry_control=base + 'translated_geometry_control.h5msm.gz',
    )
    source_files = {str(inputs_path): sha(inputs_path)}
    for role, entry in entries.items():
        assert entry['repository_path'] == paths[role]
        repository_file = (root / entry['repository_path']).resolve()
        assert repository_file.is_relative_to(root)
        stored = repository_file.read_bytes()
        assert (
            len(stored) == entry['stored_bytes']
            and hashlib.sha256(stored).hexdigest() == entry['stored_sha256']
        )
        decoded = gzip.decompress(stored)
        assert (
            len(decoded) == entry['decoded_bytes']
            and hashlib.sha256(decoded).hexdigest() == entry['decoded_sha256']
        )
        staged = Path(entry['path']).resolve()
        assert staged.is_file() and not Path(entry['path']).is_symlink()
        assert (
            staged.stat().st_size == entry['decoded_bytes']
            and sha(staged) == entry['decoded_sha256']
        )
        source_files[str(repository_file)] = entry['stored_sha256']
        source_files[str(staged)] = entry['decoded_sha256']
    assert (
        entries['result']['stored_sha256']
        == '4b1f51cdeb135fd6785a181f6bb09a0d0f631ec3d9de7c169c88703748924249'
    )
    assert (
        entries['result']['decoded_sha256']
        == 'ed0eeb3ade25e54f8a5829729578fcbec9f0d73d6c0ceb03d146b0c3492d7fac'
    )
    association = json.loads(Path(entries['association']['path']).read_text())
    assert association['schema'] == 'bindingmt.3ptb_interaction_association@1'
    assert (
        association['original_pose_count'] == 5
        and association['selected_original_position'] == 1
    )
    assert (
        association['actual_pose_atoms_preserved'] == 13
        and association['materialized_nonpolar_H'] == 5
    )
    assert association['existing_ligand_atom_correspondence'] == [
        [i, i] for i in range(13)
    ]
    for case in ('retained_pose', 'translated_geometry_control'):
        assert association['merged_ligand_atom_indices'][case] == list(
            range(3220, 3238)
        )
        assert (
            entries[case]['decoded_sha256']
            == association['files'][case + '.h5msm']['sha256']
        )
    assert not any(name == 'vina' or name.startswith('vina.') for name in sys.modules)
    assert os.sched_getaffinity(0) == {0} and os.getpriority(os.PRIO_PROCESS, 0) >= 10
    mem = next(
        x.split()[1]
        for x in Path('/proc/meminfo').read_text().splitlines()
        if x.startswith('MemAvailable:')
    )
    assert int(mem) * 1024 >= 8 * 1024**3
except BaseException as error:
    save_json(
        output / 'receipt.json',
        dict(
            owner='uibcdf/bindingmt#80',
            outcome='FAIL',
            profile=profile,
            stage=stage,
            error_type=type(error).__name__,
            reason=str(error),
            counts=counts,
        ),
    )
    raise
```

```python
class NoVinaImports:
    def find_spec(self, fullname, path=None, target=None):
        if fullname == 'vina' or fullname.startswith('vina.'):
            forbidden_attempts.append('import:' + fullname)
            raise AssertionError('Vina is outside saved-carrier receiving')
        return None


import_guard = NoVinaImports()
sys.meta_path.insert(0, import_guard)
import pyunitwizard as puw  # noqa: E402 - optional-engine guard must precede imports

caller_report = json.loads(json.dumps(puw.configure.report()))
authority_keys = ('default_form', 'default_parser', 'standard_units', 'provenance')


def authority():
    return {key: puw.configure.report()[key] for key in authority_keys}


caller_authority = authority()
units = (
    [
        'nm',
        'ps',
        'kelvin',
        'mole',
        'dalton',
        'elementary_charge',
        'kJ/mol',
        'kJ/(mol*nm)',
        'kJ/(mol*nm**2)',
        'radians',
    ]
    if profile == 'normal'
    else [
        'pm',
        'fs',
        'kelvin',
        'mole',
        'dalton',
        'coulomb',
        'kJ/mol',
        'kJ/(mol*pm)',
        'kJ/(mol*pm**2)',
        'radians',
    ]
)
unit_context = puw.context(standard_units=units)
unit_context.__enter__()
active_authority = authority()
try:
    from molsysviewer.addons import addons as viewer_addons

    counts['registry_clear'] += 1
    viewer_addons.clear()  # Publicly disables discovery in this fresh owned process.
    assert viewer_addons.names() == []
    import molsysmt as msm
    import molsysviewer as msv
    import numpy as np

    import bindingmt as dmt

    assert authority() == active_authority
except BaseException as error:
    unit_context.__exit__(None, None, None)
    sys.meta_path.remove(import_guard)
    save_json(
        output / 'receipt.json',
        dict(
            owner='uibcdf/bindingmt#80',
            outcome='FAIL',
            profile=profile,
            stage='imports',
            error_type=type(error).__name__,
            reason=str(error),
            counts=counts,
            forbidden_attempts=forbidden_attempts,
            caller_authority_restored=authority() == caller_authority,
        ),
    )
    raise


def diagnostic(value):
    """Diagnostic records only; never a molecular/interaction restoration codec."""
    # Treat JSON scalars as data before consulting quantity-form discovery.
    # In particular, the ordinary policy label 'pint' must remain plain text.
    if isinstance(value, float) and not math.isfinite(value):
        return dict(nonfinite=repr(value))
    if value is None or isinstance(value, (str, int, float, bool)):
        return value
    if isinstance(value, np.ndarray):
        return dict(
            array=diagnostic(value.tolist()),
            dtype=str(value.dtype),
            shape=list(value.shape),
        )
    if isinstance(value, np.generic):
        return diagnostic(value.item())
    if isinstance(value, dict):
        return {str(k): diagnostic(v) for k, v in value.items()}
    if isinstance(value, (list, tuple)):
        return [diagnostic(v) for v in value]
    if isinstance(value, bytes):
        return dict(bytes_hex=value.hex())
    if type(value).__name__ in ('NAType', 'NaTType'):
        return dict(missing=type(value).__name__)
    if puw.is_quantity(value):
        return dict(
            value=diagnostic(puw.get_value(value)), unit=str(puw.get_unit(value))
        )
    raise TypeError('Unsupported diagnostic value: ' + type(value).__name__)


assert diagnostic(dict(default_form='pint', value=1.0, labels=['nm', 'ps'])) == dict(
    default_form='pint', value=1.0, labels=['nm', 'ps']
)


def save(path, value):
    save_json(path, diagnostic(value))


def direct(key, function, *args, **kwargs):
    global stage
    stage = key
    assert counts[key] < budgets[key], key
    counts[key] += 1
    journal('started', direct_operation=key)
    value = function(*args, **kwargs)
    journal('returned', direct_operation=key)
    return value


def patch(owner, name, replacement):
    original = getattr(owner, name)
    patches.append((owner, name, original))
    setattr(owner, name, replacement)


def forbid(name):
    def denied(*args, **kwargs):
        forbidden_attempts.append(name)
        journal('forbidden', forbidden_operation=name)
        raise AssertionError('Forbidden saved-carrier operation: ' + name)

    return denied


def observe(function, key):
    @wraps(function)
    def call(*args, **kwargs):
        observed[key] += 1
        return function(*args, **kwargs)

    return call


for owner, name in (
    (dmt, 'dock'),
    (dmt, 'dock_many'),
    (dmt, 'score'),
    (dmt, 'prepare_ligand'),
    (dmt, 'prepare_receptor'),
    (dmt, 'evaluate_redocking'),
    (dmt, 'compare_scores'),
    (dmt, 'view'),
    (dmt, 'show'),
    (msm.build, 'add_missing_hydrogens'),
    (msm.build, 'assign_partial_charges'),
    (msm.build, 'assign_autodock_atom_types'),
    (msm.interactions.hbonds, 'get_hbonds'),
    (msm.interactions.hydrophobic, 'get_hydrophobic_interactions'),
):
    patch(owner, name, forbid(name))
for name in (
    'get_rmsd',
    'get_symmetry_corrected_rmsd',
    'get_distances',
    'get_dihedral_angles',
    'align',
    'superpose',
    'get_superposition',
):
    if hasattr(msm.structure, name):
        patch(msm.structure, name, forbid('structure.' + name))
patch(msm, 'convert', observe(msm.convert, 'public_convert'))
patch(msm, 'get', observe(msm.get, 'public_get'))
patch(msm.h5msm, 'read_layers', observe(msm.h5msm.read_layers, 'public_layer_reads'))


def payload(analysis):
    return direct(
        'typed_exports', msm.convert, analysis, to_form='molsysmt.InteractionsDict'
    ).to_dict()


def native_state(model):
    assert counts['native_snapshots'] < budgets['native_snapshots']
    counts['native_snapshots'] += 1
    coordinates = puw.get_value(msm.get(model, coordinates=True), to_unit='angstrom')
    assert coordinates.shape == (1, 3238, 3) and np.isfinite(coordinates).all()
    fields = msm.get(
        model, box=True, time=True, structure_id=True, output_type='dictionary'
    )
    box, time, structure_id = (fields[key] for key in ('box', 'time', 'structure_id'))
    return diagnostic(
        dict(
            n_atoms=int(msm.get(model, n_atoms=True)),
            n_structures=int(msm.get(model, n_structures=True)),
            topology=msm.convert(model.topology, to_form='molsysmt.TopologyDict').data,
            chemical_states=msm.convert(
                model.chemical_states, to_form='molsysmt.ChemicalStatesDict'
            ).to_dict(),
            mechanics=model.molecular_mechanics.to_dict(),
            coordinates=dict(value=coordinates, unit='angstrom'),
            box=None
            if box is None
            else dict(value=puw.get_value(box, to_unit='angstrom'), unit='angstrom'),
            time=None
            if time is None
            else dict(value=puw.get_value(time, to_unit='ps'), unit='ps'),
            structure_id=structure_id,
            structure_chemical_state_index=msm.get(
                model, structure_chemical_state_index=True
            ),
        )
    )


refusals = []


def expect_refusal(label, function, *args, expected, **kwargs):
    try:
        value = function(*args, **kwargs)
    except MemoryError:
        raise
    except Exception as error:
        record = dict(
            label=label,
            case=current,
            type=type(error).__name__,
            module=type(error).__module__,
            code=getattr(error, 'code', None),
            reason=str(error),
        )
        refusals.append(record)
        save(output / f'refusal_{len(refusals)}.json', record)
        assert type(error).__name__ in expected
        return
    raise AssertionError(
        'Expected refusal did not occur: ' + label + ' (' + type(value).__name__ + ')'
    )
```

```python
def receive_case(case, original):
    directory = output / case
    directory.mkdir()
    path = entries[case]['path']
    carrier = direct('carrier_reads', msm.convert, path, to_form='molsysmt.MolSys')
    before = native_state(carrier)
    save(directory / 'carrier_before.json', before)
    assert before['n_atoms'] == 3238 and before['n_structures'] == 1
    families = ('hbonds', 'hydrophobic')
    assert set(carrier.interactions) == set(families)
    expected = {
        family: diagnostic(payload(carrier.interactions[family])) for family in families
    }
    save(directory / 'original_full_analyses.json', expected)
    ligand = association['merged_ligand_atom_indices'][case]
    receptor = direct(
        'source_selections', msm.select, carrier, selection="molecule_type=='protein'"
    )
    np.testing.assert_array_equal(receptor, np.arange(3220))
    assert set(receptor).isdisjoint(ligand) and len(receptor) + len(ligand) == 3238
    original_indices = [
        ligand[pair[1]] for pair in association['existing_ligand_atom_correspondence']
    ]
    pose = original['poses'][0]
    for field, attribute in dict(
        atom_id='atom_id',
        atom_name='atom_name',
        element='atom_type',
        group_id='group_id',
        group_name='group_name',
    ).items():
        values = msm.get(
            carrier, element='atom', selection=original_indices, **{attribute: True}
        )
        assert [str(value) for value in values] == [
            key[field] for key in pose['metadata']['source_atom_keys']
        ]
    actual = puw.get_value(
        msm.get(carrier, selection=original_indices, coordinates=True),
        to_unit='angstrom',
    )[0]
    old = puw.get_value(
        puw.quantity(pose['coordinates']['value'], pose['coordinates']['unit']),
        to_unit='angstrom',
    )
    if case == 'retained_pose':
        np.testing.assert_allclose(actual, old, rtol=0, atol=1e-12)
    else:
        assert not np.allclose(actual, old, rtol=0, atol=1e-12)
    save(
        directory / 'original_pose_association.json',
        dict(
            original_result_sha256=entries['result']['decoded_sha256'],
            association_sha256=entries['association']['decoded_sha256'],
            original_pose_count=5,
            selected_original_position=1,
            pose_id=pose['pose_id'],
            partner_state_id=pose['partner_state_id'],
            receptor_state_id=pose['receptor_state_id'],
            original_atom_axis_to_ligand=association[
                'existing_ligand_atom_correspondence'
            ],
            ligand_to_carrier=ligand,
            admitted_original_pose_to_carrier=original_indices
            if case == 'retained_pose'
            else None,
            control_atom_axis_to_carrier=original_indices
            if case != 'retained_pose'
            else None,
            original_pose_coordinates=dict(value=old, unit='angstrom'),
            received_coordinates=dict(value=actual, unit='angstrom'),
            original_pose_coordinates_match=case == 'retained_pose',
            coordinate_tolerance=dict(value=1e-12, unit='angstrom'),
            analysis_is_original_pose=case == 'retained_pose',
            other_original_poses_unevaluated=4,
        ),
    )
    detached_carrier = direct('copies', msm.copy, carrier)
    view = direct('viewer_factory', msv.new_view)
    views.append(view)
    closed = False
    try:
        returned = direct('viewer_load', view.load, detached_carrier, mode='replace')
        assert returned is None and view.addons.available() == []
        assert native_state(view.molsys) == before
        assert {row['name'] for row in view.interactions.analyses()} == set(families)
        layer = direct(
            'independent_layer_reads',
            msm.h5msm.read_layers,
            path,
            layers=['interactions'],
            analysis_names=list(families),
        )
        assert set(layer['interactions']) == set(families)
        queries = {}
        for family in families:
            analysis = direct('analysis_get', view.interactions.get_analysis, family)
            loaded_payload = payload(analysis)
            save(directory / (family + '_loaded.json'), loaded_payload)
            assert diagnostic(loaded_payload) == expected[family]
            assert (
                diagnostic(payload(layer['interactions'][family])) == expected[family]
            )
            assert analysis.n_atoms == 3238 and analysis.n_structures == 1
            assert analysis.n_interactions == association['analyses'][case][family]
            np.testing.assert_array_equal(analysis.evaluated_structure_indices, [0])
            method, site = (
                ('baker_hubbard', 'elemental_nitrogen_oxygen')
                if family == 'hbonds'
                else ('atom_pair_distance', 'smarts_hydrophobic_atoms')
            )
            assert (
                analysis.parameters['method'] == method
                and analysis.parameters['site_definition'] == site
            )
            assert analysis.parameters['pbc'] is False
            queries[family] = direct(
                'analysis_queries',
                analysis.between_selections,
                ligand,
                receptor,
                exclusive=True,
            )
            assert queries[family].n_interactions == analysis.n_interactions
            save(
                directory / (family + '_query.json'),
                direct('query_inspections', queries[family].to_dict),
            )
            save(
                directory / (family + '_page.json'),
                direct(
                    'analysis_pages',
                    queries[family].to_page,
                    offset=0,
                    limit=20,
                    max_participant_atoms=512,
                ),
            )
            full = direct(
                'attach',
                view.interactions.attach,
                layer['interactions'][family],
                name='received_' + family,
                assume_aligned=True,
            )
            full_payload = payload(full)
            save(directory / (family + '_attached.json'), full_payload)
            assert diagnostic(full_payload) == expected[family]
            loaded = direct(
                'viewer_layer_load',
                view.interactions.load,
                path,
                analysis_name=family,
                name='loaded_' + family,
                assume_aligned=True,
            )
            independent_payload = payload(loaded)
            save(directory / (family + '_independent_load.json'), independent_payload)
            assert diagnostic(independent_payload) == expected[family]
        names = (
            set(families)
            | {'received_' + f for f in families}
            | {'loaded_' + f for f in families}
        )
        assert {row['name'] for row in view.interactions.analyses()} == names
        hbonds = layer['interactions']['hbonds']
        expect_refusal(
            'alignment_not_declared',
            direct,
            'attach',
            view.interactions.attach,
            hbonds,
            name='alignment_missing',
            expected={'InteractionAnalysisError'},
        )
        expect_refusal(
            'query_as_full',
            direct,
            'attach',
            view.interactions.attach,
            queries['hbonds'],
            name='query_not_full',
            assume_aligned=True,
            expected={'ValueError'},
        )
        expect_refusal(
            'name_conflict',
            direct,
            'attach',
            view.interactions.attach,
            hbonds,
            name='hbonds',
            assume_aligned=True,
            expected={'InteractionAnalysisError'},
        )
        wrong = direct(
            'wrong_axes_remap', hbonds.remap, atom_indices=ligand, structure_indices=[0]
        )
        save(directory / 'wrong_axes_result.json', payload(wrong))
        expect_refusal(
            'wrong_axes',
            direct,
            'attach',
            view.interactions.attach,
            wrong,
            name='wrong_axes',
            assume_aligned=True,
            expected={'ValueError'},
        )
        if case == 'retained_pose':
            for label, indices in (
                ('duplicate_atom_indices', [0, 0]),
                ('out_of_range_atom', [3238]),
            ):
                expect_refusal(
                    label,
                    direct,
                    'invalid_axes_remap',
                    hbonds.remap,
                    atom_indices=indices,
                    expected={'ValueError', 'ArgumentError'},
                )
        assert {row['name'] for row in view.interactions.analyses()} == names
        destination = directory / 'received_analyses.h5msm'
        persisted = direct(
            'viewer_save',
            view.interactions.save,
            str(destination),
            analysis_names=sorted(names),
            overwrite=False,
        )
        assert Path(persisted).resolve() == destination.resolve()
        save(
            directory / 'persistence_identity.json',
            dict(
                path=destination.name,
                bytes=destination.stat().st_size,
                sha256=sha(destination),
                geometry_included=False,
                carrier_input_sha256=entries[case]['decoded_sha256'],
            ),
        )
        reread = direct(
            'persistence_layer_reads',
            msm.h5msm.read_layers,
            str(destination),
            layers=['interactions'],
            analysis_names=sorted(names),
        )['interactions']
        assert set(reread) == names
        for name in sorted(names):
            family = name.removeprefix('received_').removeprefix('loaded_')
            restored = payload(reread[name])
            save(directory / (name + '_persisted.json'), restored)
            assert diagnostic(restored) == expected[family]
        for family in families:
            assert diagnostic(payload(carrier.interactions[family])) == expected[family]
        for name in sorted(names):
            family = name.removeprefix('received_').removeprefix('loaded_')
            final_analysis = direct(
                'analysis_get', view.interactions.get_analysis, name
            )
            assert diagnostic(payload(final_analysis)) == expected[family]
        assert native_state(view.molsys) == before
        assert native_state(carrier) == before
        assert set(carrier.interactions) == set(families)
        save(
            directory / 'case_receipt.json',
            dict(
                status='PASS',
                carrier_atoms=3238,
                carrier_structures=1,
                source_file_sha256=entries[case]['decoded_sha256'],
                families={
                    family: association['analyses'][case][family] for family in families
                },
                evaluated_structure_indices=[0],
                names=sorted(names),
                alias_names_are_new_observations=False,
                original_carrier_and_analyses_unchanged=True,
                view_molecular_state_unchanged=True,
                original_pose_binding_admitted=case == 'retained_pose',
                provenance_role='retained original pose'
                if case == 'retained_pose'
                else 'saved translated control; not another original pose',
                complete_persistence='Provider interactions-only H5MSM; source carrier remains separately authoritative',
                addon_admission=False,
                canvas_admission=False,
            ),
        )
        return dict(case=case, state=before, case_receipt=case + '/case_receipt.json')
    finally:
        primary = sys.exc_info()[1]
        try:
            direct('viewer_close', view.close)
            closed = True
        except BaseException as cleanup_error:
            if primary is None:
                raise
            primary.add_note('View cleanup failed: ' + str(cleanup_error))
        finally:
            views.remove(view)
            cleanup = dict(case=case, close_attempted=True, closed=closed)
            view_cleanup.append(cleanup)
            save(directory / 'view_cleanup.json', cleanup)
```
